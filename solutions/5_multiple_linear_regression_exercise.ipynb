**(g) Remove a couple of explanatory variables. How does $R^2$ change?**

```python
X_3 = df.drop(columns=["Sales", "US", "ShelveLoc", "Urban", "Population"])
y_3 = df["Sales"]


# Create the model
model3 = LinearRegression()

# Fit the model
model3.fit(X_3, y_3)

# Predict
predictions = model3.predict(X_3)

R_3 = model3.score(X_3, y_3)
print(f"R^2: {R_3}")
```

**(h) Repeat the process for the adjusted $R^2$.**

R²          — rewards you for adding ANY variable
              even useless ones

Adjusted R² — penalises you for adding useless variables
              only goes up if the variable is actually helpful

```python
def adjusted_r_squared(r_squared, X):
    adjusted_r2 = 1 - ((1 - r_squared) * (len(X) - 1) / (len(X) - X.shape[1] - 1))
    return adjusted_r2
```

```python
adj_R_1 = adjusted_r_squared(R_1, X_1)
adj_R_2 = adjusted_r_squared(R_2, X_2)
adj_R_3 = adjusted_r_squared(R_3, X_3)

print(adj_R_1, adj_R_2, adj_R_3)
```

## Follow-up: Adding the Categorical Variables

The models above dropped the categorical columns (`ShelveLoc`, `Urban`, `US`). As a follow-up to notebook 6, we bring them back with `pd.get_dummies`, the simplest way to turn categories into 0/1 columns a linear model can use. We pass `drop_first=True` to avoid the dummy variable trap.

```python
# Turn the three categorical columns into 0/1 dummy columns
dummies = pd.get_dummies(df[["ShelveLoc", "Urban", "US"]], drop_first=True)

# Add the dummies to the numeric predictors from the full model (X_2)
X_cat = pd.concat([X_2, dummies], axis=1)
y_cat = df["Sales"]

model_cat = LinearRegression()
model_cat.fit(X_cat, y_cat)

R_cat = model_cat.score(X_cat, y_cat)
adj_R_cat = adjusted_r_squared(R_cat, X_cat)
print(f"R^2 with categoricals:          {R_cat:.4f}")
print(f"Adjusted R^2 with categoricals: {adj_R_cat:.4f}")
```

Adding the categorical variables raises the R^2 from about 0.54 to about 0.87, a big jump. Almost all of that gain comes from `ShelveLoc`: a Good shelving location is linked to much higher sales. This is the lesson from notebook 6, categorical variables can carry a lot of signal, and `pd.get_dummies` makes them easy to feed into a linear model.
